```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Open Jupyter from this 2026 Buildathon folder.
ROOT = Path.cwd()
DATA = ROOT/'data'
OUTPUTS = ROOT/'outputs'
assert (DATA/'drought_monthly.csv').exists(), 'Open Jupyter from 2026 Buildathon.'
OUTPUTS.mkdir(exist_ok=True)
```

## Check the bundled files

```python
import hashlib
import json
checksums = json.loads((DATA/'teaching_checksums.json').read_text())
for filename, expected in checksums.items():
    actual = hashlib.sha256((DATA/filename).read_bytes()).hexdigest()
    assert actual == expected, f'File changed: {filename}'
print('Both teaching CSVs match the prepared copies.')
# Identity checks do not establish scientific validity or permission to share.
```

```python
drought = pd.read_csv(DATA/'drought_monthly.csv', parse_dates=['date'])
display(drought.head())
```

```python
example = drought[drought['division'] == 7].dropna(subset=['pdsi']).tail(24)
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(example['date'], example['pdsi'])
ax.set(title='NOAA PDSI: South Dakota division 7, last 24 valid months in snapshot',
       xlabel='Date', ylabel='PDSI (dimensionless)')
fig.tight_layout()
fig.savefig(OUTPUTS/'readiness.png')
plt.show()
print('Saved outputs/readiness.png')
```
